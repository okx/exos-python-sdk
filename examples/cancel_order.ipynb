# Cancel a signed order

Required environment variables: `EXCHANGEOS_API_KEY`, `EXCHANGEOS_API_SECRET`, `EXCHANGEOS_API_PASSPHRASE`, `EXCHANGEOS_AGENT_PRIVATE_KEY`, `EXCHANGEOS_ASSET_ID`, and exactly one of `EXCHANGEOS_ORDER_ID` or `EXCHANGEOS_CLIENT_ORDER_ID`.

Optional: `EXCHANGEOS_MARKET_TYPE` (`spot` or `perp`).

**This notebook submits a real cancellation.**

```python
import os

from exchangeos_sdk import Cancel, Credentials, ExchangeOSClient, MarketType, cancel_item

credentials = Credentials(
    api_key=os.environ["EXCHANGEOS_API_KEY"],
    secret_key=os.environ["EXCHANGEOS_API_SECRET"],
    passphrase=os.environ["EXCHANGEOS_API_PASSPHRASE"],
)
order_id = os.getenv("EXCHANGEOS_ORDER_ID")
client_order_id = os.getenv("EXCHANGEOS_CLIENT_ORDER_ID")
if (order_id is None) == (client_order_id is None):
    raise ValueError("set exactly one order identifier")
```

```python
cancel = cancel_item(
    asset_id=os.environ["EXCHANGEOS_ASSET_ID"],
    market_type=MarketType(os.getenv("EXCHANGEOS_MARKET_TYPE", "perp")),
    oid=order_id,
    client_order_id=client_order_id,
)
with ExchangeOSClient(auth=credentials) as client:
    signed_transaction = client.sign_action(
        Cancel(cancels=[cancel]),
        os.environ["EXCHANGEOS_AGENT_PRIVATE_KEY"],
    )
    result = client.trade.cancel_order(signed_transaction)

print(result)
```
