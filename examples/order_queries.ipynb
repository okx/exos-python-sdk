# Order queries

Required environment variables: `EXCHANGEOS_API_KEY`, `EXCHANGEOS_API_SECRET`, `EXCHANGEOS_API_PASSPHRASE`, `EXCHANGEOS_ASSET_ID`.

Optional: `EXCHANGEOS_VAULT_ADDRESS`.

Trade query responses use the ExchangeOS `data[0]` shape. Pending orders expose `block_height`, `block_timestamp`, and `.data`; order history and fills additionally expose `.has_more` and `.next_cursor`.

```python
import os

from exchangeos_sdk import Credentials, ExchangeOSClient

credentials = Credentials(
    api_key=os.environ["EXCHANGEOS_API_KEY"],
    secret_key=os.environ["EXCHANGEOS_API_SECRET"],
    passphrase=os.environ["EXCHANGEOS_API_PASSPHRASE"],
)
asset_id = os.environ["EXCHANGEOS_ASSET_ID"]
vault_address = os.getenv("EXCHANGEOS_VAULT_ADDRESS")
```

```python
with ExchangeOSClient(auth=credentials) as client:
    pending = client.trade.orders_pending(asset_id=asset_id, vault_address=vault_address)
    history = client.trade.orders_history(asset_id=asset_id, limit=50, vault_address=vault_address)
    fills = client.trade.fills(asset_id=asset_id, limit=50, vault_address=vault_address)

pending_page = pending[0]
history_page = history[0]
fills_page = fills[0]
print({
    "pendingOrders": len(pending_page.data),
    "historicalOrders": len(history_page.data),
    "fills": len(fills_page.data),
})
```
