# Account queries

Required environment variables: `EXCHANGEOS_API_KEY`, `EXCHANGEOS_API_SECRET`, `EXCHANGEOS_API_PASSPHRASE`.

Optional: `EXCHANGEOS_ASSET_ID`, `EXCHANGEOS_VAULT_ADDRESS`.

```python
import os

from exos_python_sdk import Credentials, ExchangeOSClient

credentials = Credentials(
    api_key=os.environ["EXCHANGEOS_API_KEY"],
    secret_key=os.environ["EXCHANGEOS_API_SECRET"],
    passphrase=os.environ["EXCHANGEOS_API_PASSPHRASE"],
)
```

```python
with ExchangeOSClient(auth=credentials) as client:
    balance = client.account.balance(vault_address=os.getenv("EXCHANGEOS_VAULT_ADDRESS"))
    positions = client.account.positions(
        asset_id=os.getenv("EXCHANGEOS_ASSET_ID"),
        vault_address=os.getenv("EXCHANGEOS_VAULT_ADDRESS"),
    )

print({"balanceEntries": len(balance), "positions": len(positions)})
```
